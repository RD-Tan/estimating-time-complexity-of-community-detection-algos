```python
from scripts.generate_barabasi_albert import generate_barabasi_albert
from itertools import product
from networkx import write_edgelist, read_edgelist
```

```python
dir_path = "../synthetic/scale-free/"
node_count_list = [1024, 2048, 4096, 8192, 16384, 32768]
m_list = [8, 16, 32, 64, 128, 256, 512]
```

```python
for n, m in product(node_count_list, m_list):
    write_edgelist(generate_barabasi_albert(n, m), f"{dir_path}/n-{n}-m-{m}.csv")
```

```python
report = []
for n, m in product(node_count_list, m_list):
    graph = read_edgelist(f"{dir_path}/n-{n}-m-{m}.csv")
    edge_count = graph.size()
    node_count = graph.order()
    density = edge_count / (node_count * (node_count - 1) // 2)
    report.append(f"nodes: {node_count}, edges: {edge_count}, m: {m}, density: {density}")
    print(f"nodes: {node_count}, edges: {edge_count}, m: {m}, density: {density}")
```
