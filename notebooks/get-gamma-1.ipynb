```python
from networkx import read_edgelist, barabasi_albert_graph
import powerlaw
import numpy as np
```

```python
def calculateGamma(filename: str):
    with open(filename) as f:
        network = read_edgelist(f)
        degrees = [d for _, d in network.degree()]

    fit = powerlaw.Fit(degrees, discrete=True)

    # 3. Extract the estimated scaling exponent (alpha / gamma)
    gamma = fit.alpha
    xmin = fit.xmin

    print(f"Estimated gamma: {gamma}")
    print(f"Lower bound xmin: {xmin}")
```

```python
calculateGamma("../actual/mapping/STRING/int-edge-list.txt")
```

```python
calculateGamma("../actual/mapping/STRING-400/int-edge-list.txt")
```
